Create two [Column Transformers](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html), called preproc1 and preproc2, with the following guidelines:

### Preproc 1

Create preproc1 below.

+ Numeric: scaled variables, no other transforms.
+ Categorical: one-hot encoding.

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Define numerical and categorical columns
numerical_cols = ['coord_x', 'coord_y', 'ffmc', 'dmc', 'dc', 'isi', 'temp', 'rh', 'wind', 'rain']
categorical_cols = ['month', 'day']

# Create the ColumnTransformer for preproc1
preproc1 = ColumnTransformer([
    ("num_scaling", StandardScaler(), numerical_cols),  # Scale numerical features
    ("cat_encoding", OneHotEncoder(handle_unknown='ignore'), categorical_cols)  # One-hot encode categorical features
])

# Display the transformer
print(preproc1)
```

### Preproc 2

Create preproc1 below.

+ Numeric: scaled variables, non-linear transformation to one or more variables.
+ Categorical: one-hot encoding.

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, PowerTransformer

# Define numerical and categorical columns
numerical_cols = ['coord_x', 'coord_y', 'ffmc', 'dmc', 'dc', 'isi', 'temp', 'rh', 'wind', 'rain']
categorical_cols = ['month', 'day']

# Select skewed numerical columns for transformation (example: 'dc' and 'isi')
skewed_cols = ['dc', 'isi']

# Create the ColumnTransformer for preproc2
preproc2 = ColumnTransformer([
    ("num_scaling", StandardScaler(), [col for col in numerical_cols if col not in skewed_cols]),  # Standard scaling
    ("num_transformation", PowerTransformer(method='yeo-johnson'), skewed_cols),  # Non-linear transformation
    ("cat_encoding", OneHotEncoder(handle_unknown='ignore'), categorical_cols)  # One-hot encoding
])

# Display the transformer
print(preproc2)
```

## Regressor

+ Use a regression model to perform a prediction. 

    - Choose a baseline regressor, tune it (if necessary) using grid search, and evaluate it using cross-validation.
    - Choose a more advance regressor, tune it (if necessary) using grid search, and evaluate it using cross-validation.
    - Both model choices are up to you, feel free to experiment.

```python
# Pipeline A = preproc1 + baseline
import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, PowerTransformer
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV, cross_val_score

# Define numerical and categorical columns
numerical_cols = ['coord_x', 'coord_y', 'ffmc', 'dmc', 'dc', 'isi', 'temp', 'rh', 'wind', 'rain']
categorical_cols = ['month', 'day']
skewed_cols = ['dc', 'isi']  # Skewed features for transformation

# Define preprocessing transformers
preproc1 = ColumnTransformer([
    ("num_scaling", StandardScaler(), numerical_cols),  
    ("cat_encoding", OneHotEncoder(handle_unknown='ignore'), categorical_cols)  
])

preproc2 = ColumnTransformer([
    ("num_scaling", StandardScaler(), [col for col in numerical_cols if col not in skewed_cols]),  
    ("num_transformation", PowerTransformer(method='yeo-johnson'), skewed_cols),  
    ("cat_encoding", OneHotEncoder(handle_unknown='ignore'), categorical_cols)  
])

# Define regression models
baseline_model = LinearRegression()
advanced_model = RandomForestRegressor(n_estimators=100, random_state=42)

# Create Pipelines
pipeline_A = Pipeline([
    ('preprocessing', preproc1),
    ('regressor', baseline_model)
])
```

```python
# Pipeline B = preproc2 + baseline
pipeline_B = Pipeline([
    ('preprocessing', preproc2),
    ('regressor', baseline_model)
])
```

```python
# Pipeline C = preproc1 + advanced model
pipeline_C = Pipeline([
    ('preprocessing', preproc1),
    ('regressor', advanced_model)
])
```

```python
# Pipeline D = preproc2 + advanced model
pipeline_D = Pipeline([
    ('preprocessing', preproc2),
    ('regressor', advanced_model)
]) 
```

# Tune Hyperparams

+ Perform GridSearch on each of the four pipelines. 
+ Tune at least one hyperparameter per pipeline.
+ Experiment with at least four value combinations per pipeline.

```python
# Define hyperparameter grids
param_grid_A = {'regressor__fit_intercept': [True, False]}
param_grid_B = {'regressor__fit_intercept': [True, False]}
param_grid_C = {'regressor__n_estimators': [50, 100, 200, 300], 'regressor__max_depth': [None, 10, 20, 30]}
param_grid_D = {'regressor__n_estimators': [50, 100, 200, 300], 'regressor__max_depth': [None, 10, 20, 30]}
```

```python
# Run GridSearchCV
pipelines = {
    "Pipeline A": (pipeline_A, param_grid_A),
    "Pipeline B": (pipeline_B, param_grid_B),
    "Pipeline C": (pipeline_C, param_grid_C),
    "Pipeline D": (pipeline_D, param_grid_D)
}

best_models = {}
```

```python
for name, (pipeline, param_grid) in pipelines.items():
    grid_search = GridSearchCV(pipeline, param_grid, cv=5, scoring='neg_mean_squared_error', n_jobs=-1)
    grid_search.fit(X, y)
    
    print(f"Best parameters for {name}: {grid_search.best_params_}")
    print(f"Best MSE for {name}: {-grid_search.best_score_:.4f}\n")
    
    best_models[name] = grid_search.best_estimator_
```
